# Vanilla 3D Autoencoder Training

Self-contained training notebook for MNIST autoencoder with 3D latent space.

```python
import importlib
import standard_3d_autoencoder as ae3d
importlib.reload(ae3d)  # Reload to get fixed architecture

# Config
CONFIG = {
    'latent_dim': 3,
    'epochs': 30,
    'lr': 1e-3,
    'batch_size_train': 128,
    'batch_size_test': 64
}
```

```python
# Setup
device = ae3d.get_device()
train_loader, test_loader = ae3d.get_mnist_loaders(CONFIG['batch_size_train'], CONFIG['batch_size_test'])
model = ae3d.StandardAutoEncoder(CONFIG['latent_dim'])
print(f"Device: {device}")
print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Train
losses = ae3d.train_autoencoder(model, train_loader, CONFIG['epochs'], CONFIG['lr'], device)
```

```python
# Evaluate
test_loss = ae3d.evaluate_model(model, test_loader, device)
ae3d.plot_training_losses(losses)
```

```python
# Visualize results
ae3d.visualize_reconstructions(model, test_loader, device)
```

```python
# 3D latent space
latent_data, labels = ae3d.visualize_3d_latent_space(model, test_loader, device)
```

```python
# Latent interpolation
generated_images = ae3d.latent_grid_interpolation(model, device, grid_size=8)
```

```python
# Save model
ae3d.save_model(model, 'autoencoder_3d.pth', CONFIG, losses)
```
